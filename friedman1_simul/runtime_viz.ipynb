```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dat_LS = pd.read_csv('vizes/runtime_LS.csv')
dat_LAD = pd.read_csv('vizes/runtime_LAD.csv')
plt.figure(figsize = (20,5))
plt.subplot(1,3,1)
sns.lineplot(data=dat_LS, x='n_samples', y='time (s)')
plt.xlabel('Target instances')
plt.ylabel('Runtime (s)')
plt.title('LS')
plt.subplot(1,3,2)
sns.lineplot(data=dat_LAD, x='n_samples', y='time (s)')
plt.xlabel('Target instances')
plt.ylabel('Runtime (s)')
plt.title('LAD')
```
