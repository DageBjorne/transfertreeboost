```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
dat_LS = pd.read_csv('vizes/runtime_LS.csv')
dat_LAD = pd.read_csv('vizes/runtime_LAD.csv')
dat_M = pd.read_csv('vizes/runtime_M.csv')
plt.figure(figsize = (20,5))
plt.subplot(1,3,1)
sns.lineplot(data=dat_LS, x='n_samples', y='time (s)', ci=None, marker = 'o')
plt.xlabel('Target instances', fontsize = 16)
plt.ylabel('Average runtime (s)', fontsize = 16)
plt.title('LS', fontsize = 20)
plt.subplot(1,3,2)
sns.lineplot(data=dat_LAD, x='n_samples', y='time (s)', ci=None, marker = 'o')
plt.xlabel('Target instances', fontsize = 16)
plt.ylabel('Average runtime (s)', fontsize = 16)
plt.title('LAD', fontsize = 20)
plt.subplot(1,3,3)
sns.lineplot(data=dat_M, x='n_samples', y='time (s)', ci=None, marker = 'o')
plt.xlabel('Target instances', fontsize = 16)
plt.ylabel('Average runtime (s)', fontsize = 16)
plt.title('M', fontsize = 20)
plt.savefig('vizes/runtime.png', bbox_inches = 'tight', pad_inches = 0.1, dpi = 500)
```
